```julia
open("demo.txt", "w") do f
    write(f, "строка 1\n")
    write(f, "строка 2\n")
    write(f, "строка 3\n")
end

println("read() -- весь файл одной строкой:")
println(read("demo.txt", String))

println("readline() -- только первая строка:")
open("demo.txt") do f
    println(readline(f))
end

println("readlines() -- вектор всех строк:")
println(readlines("demo.txt"))
```
